# 🚀 Ephemeral LLM Runner on Kaggle

Run open-weight frontier LLMs (**DeepSeek-R1 Distill**, **Qwen 2.5**, **Phi-4**, **Llama 3.1**) on Kaggle's free GPU (2x T4 or P100) and access it remotely from VS Code or local apps via a **free Ngrok static domain** (no credit card required).

### Prerequisites:
1. **GPU Accelerator:** Under *Notebook options* (right panel) -> *Accelerator* -> Choose **GPU T4 x2** or **GPU P100**.
2. **Internet:** Toggle **Internet: ON** in the right panel.
3. **Ngrok Secrets (Free, No Card):** In top menu: *Add-ons* -> *Secrets*:
   - `NGROK_AUTHTOKEN`: (From your free [ngrok.com](https://ngrok.com) account)
   - `NGROK_DOMAIN`: (Your free static domain, e.g. `cricket-bold-regularly.ngrok-free.app`)

```python
# Step 1: Load Ngrok credentials from Kaggle Secrets
import os
try:
    from kaggle_secrets import UserSecretsClient
    secrets = UserSecretsClient()
    os.environ["NGROK_AUTHTOKEN"] = secrets.get_secret("NGROK_AUTHTOKEN")
    os.environ["NGROK_DOMAIN"] = secrets.get_secret("NGROK_DOMAIN")
    print(f"✔ Ngrok credentials loaded! Target static domain: https://{os.environ['NGROK_DOMAIN']}")
except Exception:
    print("ℹ No Kaggle secrets configured. Running locally inside the notebook.")
```
